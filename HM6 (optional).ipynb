```python
import json #Импортируем json
```

```python
purshaces={} #Создаем пустой словарь покупок
```

```python
with open('purchase_log.txt',encoding='utf-8') as f: #Открываем файл
    for line in f: #перебираем строки в файле
        line=line.strip() #Убираем лишние пробелы и переносы строк
        dict_=json.loads(line) #Переводим лог покупок в формат json 
        if dict_['user_id'] != 'user_id': #Пропускаем заголовок в файле
            purshaces[dict_['user_id']] = dict_['category'] #Записываем значения из файла в пустой словарь
```

```python
with open('visit_log.csv',encoding='utf-8') as f1, \ 
open('funnel.csv', 'w', encoding='utf-8-sig') as output: #Открываем файлы для работы с ними
    for line in f1: #Перебираем строки 
        user_id,source=line.strip().split(',') #Распиливаем через запятую и присваиваем две переменные
        if user_id=='user_id': #Пропускаем заголовки столбцов
            continue #Продолжаем цикл
        if user_id in purshaces: #Условие наличия user ID из visitlog в словаре purchase_log
           category=purshaces[user_id] #Добавляем переменную категория и присваиваем ей значения из словаря по типу покупки
           output.write(user_id+','+source+','+category+'\n') #Записываем в файл изменения
           
```
